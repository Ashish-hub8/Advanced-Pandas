Q4 Regional Sales Performance

Calculate the total sales generated by each region.

Then identify:
*   Which region generated the highest revenue?




```python
region_sales = df.groupby("Region")["Sales"].sum()

print(region_sales)

print("Highest Revenue Region:",region_sales.idxmax())
```

Q7 Pivot Table Creation
Create a pivot table showing:

Q8 Profitability Analysis

Q10 End-to-End Data Wrangling Task

Perform the following workflow:

1.   Filter the dataset where Sales > 3000
2.   Group data by Region and Product
3.   Calculate total sales
4.   Reshape the result into a pivot table

Explain the business insights obtained from this analysis.

```python
# Step 1: Filter the data
filtered = df[df["Sales"] > 3000]

# Step 2: Group by Region and Product and calculate Total Sales
grouped = filtered.groupby(["Region", "Product"])["Sales"].sum().reset_index()

# Display Total Sales
print("Total Sales by Region and Product:")
print(grouped)

# Step 3: Convert into Pivot Table
pivot_table = grouped.pivot(
    index="Region",
    columns="Product",
    values="Sales"
)

# Display Pivot Table
print("\nPivot Table:")
print(pivot_table)
```

**Business Insights**

*   The analysis includes only high-value orders (Sales > 3000).
*   The pivot table shows the total sales of each product across different regions.
*   East contributes significantly to high-value sales, especially for Laptops.
*   The results help identify the best-performing products in each region and support decisions related to inventory planning, sales strategy, and regional marketing.
